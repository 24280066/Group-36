```python
!pip install praw great_expectations pandas

import praw
import pandas as pd
import great_expectations as gx

#Fetch Data from Reddit
reddit = praw.Reddit(
    client_id="0Eb1uAXqkGUZiaK24o235A",
    client_secret="TGv4CccggfKoRfKW9Z1k4SMxvuKUUg",
    user_agent="GX_Lab_3_Scraper/1.0 (by u/Acrobatic_East_6708)"
)

#Extracting Data from "technology" Subreddit
posts = []
subreddit = reddit.subreddit("technology")
for post in subreddit.hot(limit=100):
    posts.append([post.title, post.score, post.id, post.url, post.num_comments, post.created])

#Data Frame
df = pd.DataFrame(posts, columns=["title", "score", "id", "url", "num_comments", "created"])

#Initialize Great Expectations Context
context = gx.get_context()

# Define Data Source & Asset
data_source = context.data_sources.add_pandas("pandas_source")
data_asset = data_source.add_dataframe_asset(name="reddit_data")

# Define Batch & Retrieve Data
batch_definition = data_asset.add_batch_definition_whole_dataframe("batch_definition")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})

# Creating Validator Directly
validator = context.get_validator(batch=batch)

# Data Quality Expectations

# Row-Level Validation: "title" column must have no null values
validator.expect_column_values_to_not_be_null(column="title")

# Column-Level Validation: "score" is in a reasonable range
validator.expect_column_values_to_be_between(column="score", min_value=0, max_value=50000)

# Multi-Column Validation: "score" should be greater than "num_comments"
validator.expect_column_pair_values_A_to_be_greater_than_B(column_A="score", column_B="num_comments")

# Table-Level Validation: Ensure table has at least 50 rows
validator.expect_table_row_count_to_be_between(min_value=50, max_value=100000)

# Distribution-Level Validation: Mean number of comments must be within range
validator.expect_column_mean_to_be_between(column="num_comments", min_value=10, max_value=10000)

# Step 12: Run Validation
validation_result = validator.validate()

# Step 13: Print Validation Results (Fixed)
print("\nValidation Success:", validation_result.success)

# Fixed the issue: Use `.expectation_config.expectation_type` correctly
for res in validation_result.results:
    expectation_type = res.expectation_config.expectation_type if hasattr(res.expectation_config, "expectation_type") else "Unknown Expectation"
    print(f"{expectation_type}: {res.success}")
```
